## 6. OOP

```python
class Person:
    
    
    def __init__(self, name, age):
        self.name = name
        self.age = age
        
        
    def greetings(self):
        print("Hello " + self.name)
```

```python
p1 = Person("Maria", 22)
p1.greetings()
```

```python
class Student(Person):
    
    studNum=0
    
    def __init__(self, name, age, matNo):
        super().__init__(name,age)
        self.__matNo = matNo
        Student.studNum += 1
        
        
    def getMatNo(self):
        return self.__matNo
```

```python
s1 = Student("Melanie", 21, 3049857)
s2 = Student("Jens", 43, 324566)
```

```python
# s1.__matNo

s1.getMatNo()
```
